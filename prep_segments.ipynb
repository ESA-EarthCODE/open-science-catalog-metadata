```python
import sys
sys.path.append('/home/potzschf/repos/')
import geopandas as gpd
from helperToolz.helper import getFilelist
```

```python
path_files = "/data/Aldhani/eoagritwin/fields/10_from_Gideon/v1/v1/segments/parquet/"
path_store = "/data/Aldhani/eoagritwin/fields/10_from_Gideon/v1/v1/segments/parquet_upload/"
```

```python
for file in getFilelist(path_files, "parquet"):
    file_name = file.split("/")[-1]
    gdf = gpd.read_parquet(file)

    gdf["ID"] = range(1, len(gdf) + 1)
    gdf = gdf.drop(columns=["dn"])

    gdf = gdf[["ID", "geometry"]]
    gdf.to_parquet(f"{path_store}{file_name}", index=False)
```
